### Step 2

For this homework, you will be working with the training set. Check out the `Train.csv`, open it and see what it contains. Load the dataset and plot random samples.

```python
# Load the training labels
root = './data'
data = pd.read_csv(os.path.join(root, 'Train.csv'))

# Number of training samples (amount of samples in data)
num_samples = len(data)

# Show random data samples
for ii in range(15):
    # Get random index
    idx = np.random.randint(0, num_samples)
    # Load image
    img = cv2.imread(os.path.join(root, data.iloc[idx]['Path']))
    # Convert image to RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Show image
    plt.subplot(3,5,ii+1), plt.imshow(img), plt.title(data.iloc[idx]['ClassId'])
```

### Step 3

Inspect the dataset by computing and plotting the per-class histogram.

```python
# Extract class identifiers
ids = data['ClassId'].values
```

Compute the per class histogram. You can use any approach you want (e.g. `numpy`). It's also worth looking at the `Counter` function from the `collections` module ([link](https://docs.python.org/3/library/collections.html#collections.Counter)) ;-)

```python
from collections import Counter
hist = Counter(ids)

plt.bar(hist.keys(), hist.values()), plt.grid(True)
plt.xlabel('Traffic Sign ID'), plt.ylabel('Counts')
```

### Optional

Perform a further analysis on the dataset and draw some conclusion from it.

Hint 1: Unlike MNIST or CIFAR10, this dataset contains images with various spatial resolutions. Is there anything we can tell about the resolution distribution?

Hint 2: What about the brightness distribution? Are there classes there are significantly more bright than others?

```python
data['Resolution'] = data['Width'] * data['Height']

plt.figure(figsize=(8, 5))
plt.hist(data['Resolution'], bins=30)
plt.xlabel('Image Resolution (Width × Height)')
plt.ylabel('Frequency')
plt.title('Distribution of Image Resolutions')
plt.show()

avg_res_per_class = data.groupby('ClassId')['Resolution'].mean()
plt.figure(figsize=(8, 5))
avg_res_per_class.plot(kind='bar')
plt.xlabel('Traffic Sign ID')
plt.ylabel('Average Resolution')
plt.title('Average Image Resolution per Class')
plt.show()
```

The dataset shows a strong variation in image resolutions, with most samples being small and a few having much higher resolutions, causing distribution to be right-skewed. Average resolution also differs notably across classes. Some traffic sign IDs have much larger images than others, indicating that certain signs were captured closer or are physically larger. This imbalance in resolution may introduce bias, causing models to rely on image size rather than true visual features. Standardizing image dimensions and applying scale-based data augmentation are essential to ensure fair and consistent model training.
